# 4-Layer IoE Architecture Simulation
This notebook models and simulates a 4-layer Internet of Everything (IoE) architecture:
1. **Perception Layer**: RFID/NFC device identification and sensor data acquisition.
2. **Network Layer**: Data transmission simulating MQTT publish/subscribe mechanisms.
3. **Middleware Layer**: Node-RED style data routing, parsing, and integration.
4. **Application Layer**: Cloud dashboard visualization of device metrics and alerts.

```python
import time
import random
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

print("Libraries successfully imported. Environment ready for IoE simulation!")
```

## 1. Perception Layer (RFID/NFC & Sensors)
We will define a set of authorized RFID/NFC device tags and simulate continuous environmental sensor data generation linked to those devices.

```python
import random
import json
from datetime import datetime

# Define static RFID/NFC device inventory
DEVICE_INVENTORY = {
    "TAG_E402_9F1A": {"device_name": "Conveyor_Sensor_A", "location": "Zone_1", "status": "Active"},
    "TAG_A811_3B8C": {"device_name": "Sorting_Arm_B", "location": "Zone_2", "status": "Active"},
    "TAG_D294_7C0E": {"device_name": "Packaging_Unit_C", "location": "Zone_3", "status": "Maintenance"}
}

def simulate_perception_layer():
    """Simulates physical device scanning and sensor readout."""
    tag_id = random.choice(list(DEVICE_INVENTORY.keys()))
    device_info = DEVICE_INVENTORY[tag_id]

    # Generate physical sensor values linked to physical tags
    telemetry = {
        "timestamp": datetime.now().isoformat(),
        "tag_id": tag_id,
        "device_name": device_info["device_name"],
        "location": device_info["location"],
        "temperature_c": round(random.uniform(18.0, 32.0), 2),
        "humidity_pct": round(random.uniform(30.0, 70.0), 2),
        "vibration_hz": round(random.uniform(5.0, 150.0), 2) if device_info["status"] == "Active" else 0.0
    }
    return telemetry

# Sample output of one perception scan
sample_data = simulate_perception_layer()
print(json.dumps(sample_data, indent=2))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Simulate generating multiple perception data points over time
telemetry_history = []
for _ in range(20):
    telemetry_history.append(simulate_perception_layer())
    time.sleep(0.05) # Brief pause to simulate time increments

# Convert history to a DataFrame
df_history = pd.DataFrame(telemetry_history)
df_history['timestamp'] = pd.to_datetime(df_history['timestamp'])

# Plot the historical trends
fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)

# Temperature plot
axes[0].plot(df_history['timestamp'], df_history['temperature_c'], color='red', marker='o')
axes[0].set_title('Sensor Telemetry History')
axes[0].set_ylabel('Temp (°C)')
axes[0].grid(True, linestyle='--', alpha=0.5)

# Humidity plot
axes[1].plot(df_history['timestamp'], df_history['humidity_pct'], color='blue', marker='s')
axes[1].set_ylabel('Humidity (%)')
axes[1].grid(True, linestyle='--', alpha=0.5)

# Vibration plot
axes[2].plot(df_history['timestamp'], df_history['vibration_hz'], color='green', marker='^')
axes[2].set_ylabel('Vibration (Hz)')
axes[2].set_xlabel('Timestamp')
axes[2].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
